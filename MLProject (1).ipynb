Feature Engenierring

```python
#Creating a copy of data
df_clean = df.copy()
```

```python
# Dropping Duration
df_clean = df_clean.drop(columns=['duration'])
```

we cannot the duration of the Call before the call

```python
# Creating a feature for people never contacted
df_clean['previously_contacted'] = np.where(
    df_clean['pdays'] == 999,0,1)
```

```python
#dropping pdays columns
df_clean = df_clean.drop(columns=['pdays'])
```

```python
# Creating Age Group Bins
bins = [17, 25, 35, 45, 55, 100]

labels = [
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56+'
]

df_clean['age_group'] = pd.cut(
    df_clean['age'],
    bins=bins,
    labels=labels
)
```

```python
# Campaign Feature
df_clean['multiple_contacts'] = np.where(
    df_clean['campaign'] > 1,1,0)
```

Encoding Categorical Variable

```python
# Import One hot Encoder
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create the Encoder
encoder = OneHotEncoder(handle_unknown="ignore",
                        sparse_output=False)
```

```python
print(cat_cols)
```

```python
# Fit on Training Data
encoded_train = encoder.fit_transform(X_train[cat_cols])
```

```python
encoded_test = encoder.transform(X_test[cat_cols])
```

```python
# Get New Columns Name
encoded_columns = encoder.get_feature_names_out(
    cat_cols)
```

```python
# New Features created with one hot encoding
encoded_columns
```

```python
# Convert into dataframe
# Training
encoded_train_df = pd.DataFrame(
    encoded_train,
    columns=encoded_columns,
    index=X_train.index)
```

```python
#Testing
encoded_test_df = pd.DataFrame(
    encoded_test,
    columns=encoded_columns,
    index=X_test.index
)
```

```python
#Remove Original Categorical Columns
X_train_num = X_train.drop(columns=cat_cols)
X_test_num = X_test.drop(columns=cat_cols)
```

```python
#Combine Numerical + Encoded Features
X_train_final = pd.concat(
    [X_train_num, encoded_train_df],
    axis=1)
```

```python
X_test_final = pd.concat(
    [X_test_num, encoded_test_df],
    axis=1)
```

```python
#Verify
print(X_train_final.shape)
print(X_test_final.shape)

print(X_train_final.head())
```

**Model Evaluation**

```python
#Accuracy
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

90% Accuracy doesnt mean the model is perfect

```python
#Confusion Matrix
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred)
print(cm)
```

- 209 where the customer subscribed and the model actualy predicted Yes (TP)
- 7218 where model correctly predicted NO (TN)
- 96 where the model thought customer would subscribe. But they didn't The bank wastes a sales call. (FP)
- 719 where the model suggest Don't Call this customer, But the Customer would have subscribed, This means bank would loose a sale (FN)

```python
# Classification Report
from sklearn.metrics import classification_report

print(classification_report(y_test,y_pred))
```

```python
# AUC
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(
    y_test,
    y_prob_yes)
print(auc)
```

AUC measures the quality of the predicted probabilities.

```python
#ROC
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob_yes,
    pos_label='yes')

plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0,1],[0,1],'--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

ROC Curve evaluates the model across many thresholds.

It plots:
True Positive Rate (Recall) on the Y-axis.
False Positive Rate on the X-axis.
A better model produces a curve that bows toward the top-left corner.

- our predicition about "no" increases that means the model is predicting excelelnt in terms of whom not to contact
- Performance drops drastically for the positive class (support: 928). While the precision is moderately okay at 71% (meaning when it predicts "yes", it is right 71% of the time), the recall is very poor at 22%. This means the model only successfully identifies 22% of all actual "yes" cases, missing the other 78% entirely (the 722 False Negatives).
